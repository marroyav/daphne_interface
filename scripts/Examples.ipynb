```python
import numpy as np
from ivtools import Daphne
from matplotlib import pyplot as plt

# Initialize Daphne object
ip_address = "10.73.137.107"  # Replace with your hardware IP
port = 2001
daphne = Daphne(ip_address, port)

# Parameters
afe = 0  # Replace with your AFE number
channel = 0  # Replace with your channel number
samples = 1000  # Number of samples per waveform
num_waveforms = 1000  # Number of waveforms to collect

# Collect waveforms
waveforms = []
for i in range(num_waveforms):
    print(f"Collecting waveform {i + 1}/{num_waveforms}")
    wf = daphne.read_waveform(afe=afe, ch=channel, samples=samples)
    if wf.size > 0:
        waveforms.append(wf)
    else:
        print(f"Waveform {i + 1} is empty or invalid.")

# Convert to NumPy array for further processing
waveforms = np.array(waveforms)

# Plot the waveforms
plt.figure(figsize=(12, 6))
for i, wf in enumerate(waveforms):  # Plot first 10 waveforms as an example
    plt.plot(np.linspace(0, len(wf[:10]) * 16e-9, len(wf)), wf, alpha=0.5, label=f"Waveform {i + 1}")

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Collected Waveforms")
plt.grid()
plt.legend()
plt.show()

# Optionally compute and plot the mean waveform
mean_waveform = np.mean(waveforms, axis=0)
plt.figure(figsize=(12, 6))
plt.plot(np.linspace(0, len(mean_waveform) * 16e-9, len(mean_waveform)), mean_waveform, linewidth=2, label="Mean Waveform")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Mean Waveform")
plt.grid()
plt.legend()
plt.show()
```

```python
from ivtools import Daphne

def test_read_current():
    IP = "10.73.137.112"  # Replace with your device IP
    CHANNEL = 20

    device = Daphne(IP)
    try:
        print("Testing current reading...")
        current = device.read_current(ch=CHANNEL)
        print(f"Current for Channel {CHANNEL}: {current} mA")
    except Exception as e:
        print(f"Failed to read current: {e}")
    finally:
        device.close()

if __name__ == "__main__":
    test_read_current()
```

```python
from ivtools import Daphne
import os
from time import sleep

def main():
    IPS = ["10.73.137.107", "10.73.137.113"]
    AFE = 1
    CHANNELS = [0, 7]
    SAMPLES = 4000
    REPEATS = 10
    DT = 16e-9
    SAVE_DIR = "plots"
    os.makedirs(SAVE_DIR, exist_ok=True)

    device = Daphne(IPS[0])
    try:
        print("Reading waveform for AFE 4, Channel 0...")
        waveform = device.read_waveform(afe=AFE, ch=0, samples=SAMPLES, plot=True, save_path=f"{SAVE_DIR}/waveform_afe4_ch0.png")

        print("Computing FFT for a single waveform...")
        fft_x, fft_y = device.compute_fft(waveform, dt=DT, plot=True, save_path=f"{SAVE_DIR}/fft_afe4_ch0.png")

        print("Computing mean FFT for multiple waveforms from AFE 4, Channel 1...")
        waveforms = [device.read_waveform(afe=AFE, ch=1, samples=SAMPLES) for _ in range(REPEATS)]
        mean_fft_x, mean_fft_y, mean_rms = device.compute_mean_fft(waveforms, label="AFE 4 CH 1", plot=True, save_path=f"{SAVE_DIR}/mean_fft_afe4_ch1.png")
        print(f"Mean FFT RMS for AFE 4, Channel 1: {mean_rms}")

        print("Reading current from Channel 0...")
        device.command("RESET")
        sleep(1)
        current = device.read_current(ch=0)
        print(f"Current for {IPS[0]}, Channel 0: {current} mA")
    except Exception as e:
        print(f"Error: {e}")
    finally:
        device.close()
        print("Device connection closed.")

if __name__ == "__main__":
    main()
```
